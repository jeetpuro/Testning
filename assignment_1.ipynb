### 2.2 Branch coverage
Now compute the branch coverage of the program using [`coverage.py`](https://coverage.readthedocs.io/en/latest/index.html). (**2 points**)

```python
!coverage run --branch test_estimate_pi.py
!coverage report -m
```

How can we interprete the results?

    Interpretation

    We could interprete branch coverage as how much of the logical paths that has been covered, not just how many lines, but also how it is covered in each statements

    Stmts: defined as any executable line of code that performs an action, such as variable assignments, function calls, or control flow statements.

    Miss: What statements that the program has not executed.

    Branch: example in an if-else statement, we have 2 branches, and if we only have gone in to the if branch and not the else, we have 2 branches but only 1 coverd. in Branch coverage we try to check all possible outcomes that the program can give.

    BrPart: How many parts of the branches we ahve found, are not checked.

    Cover: percentage of the code that have been executed by the program.

    Missing: the Rows that has not been executed.

    91% coverage in test_estimate_pi.py. meaning that some Statements in "test_estimate_pi.py" runs.

    53% coverage in "estimate_pi.py" because our single test in "test_estimate_pi.py" only runs/calls a fraction of the code/statements in "estimate_pi.py". Why it droped form 54% to 53%, is because we now ahve more things we check for. Mainly beacause the program jumps over 1 possible path "BrPart" in a branch and this is the cause of the decrease.

    26 Total statments in estimate_pi.py, 12 of them were not exectued. the 12 rows that were missing are: 18-19, 23-36. additionally we have 10 branches, and 1 branch that was only partially executed.

    test_estimate_pi.py have 2 branches. 1 branch was partially executed.

    Lastly we have the "total" row where it summerize the statements, miss, coverage and missing across the files that we ran.

## 4. Mocking

We want to store the resulting number persistently on our file system. We use the following class. 

```python
class PiFileWriter:
    @staticmethod
    def write(content, file_path):
        with open(file_path, 'w') as file:
            file.write(content)
```
